````python
def set_one_two(d):  # {4:8, 2:9}
    d[1] = 2  # {4:8, 2:9, 1:2}
    print("inside", d)  # inside {4:8, 2:9, 1:2}


x = {4: 8, 2: 9}


set_one_two(x)
print("outside", x)  # outside {4:8, 2:9, 1:2}
````

````python
import copy


def set_one_two(d):
    d[1] = 2
    print("inside", d)


x = {4: 8, 2: 9}
new_x = copy.deepcopy(x)

set_one_two(new_x)
print("outside", x)
````
