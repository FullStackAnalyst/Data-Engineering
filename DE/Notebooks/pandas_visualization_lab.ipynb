### 2. Loading the Dataset

```python
import pandas as pd
import seaborn as sns

# Load the classic iris dataset
df = sns.load_dataset('iris')
df.head()
```

---
## 1. Scatter Plot
**What it is:** A plot that uses dots to represent values for two different numeric variables.

**When to use:** Use a scatter plot to observe relationships, correlations, or distributions between two continuous numeric features (e.g., checking if longer petals are wider).

```python
df.plot.scatter(
    x='sepal_length',
    y='sepal_width',
    title='Sepal Length vs Sepal Width',
    grid=True
)
```

---
## 2. Line Chart
**What it is:** A plot that displays information as a series of data points connected by straight line segments.

**When to use:** Use line charts to track changes over periods of time or across ordered sequential records. Here, we track the trend of measurement values across continuous observations.

```python
df[['petal_length', 'petal_width']].plot.line(
    title='Petal Measurements Sequential Trend',
    figsize=(10, 4)
)
```

---
## 3. Bar Chart
**What it is:** A plot that presents categorical data with rectangular bars with heights or lengths proportional to the values they represent.

**When to use:** Use to compare numerical metrics across different distinct categories or groups. Here, we aggregate the dataset to compare the average sepal length across species.

```python
species_mean = df.groupby('species')['sepal_length'].mean()
species_mean.plot.bar(
    title='Average Sepal Length by Species',
    ylabel='Length (cm)',
    rot=0
)
```

---
## 4. Histogram
**What it is:** An accurate representation of the distribution of numerical data by binning consecutive ranges of data.

**When to use:** Use when you want to understand the underlying frequency distribution, spread, skewness, or potential outliers of a continuous numerical attribute.

```python
df['petal_length'].plot.hist(
    bins=15,
    title='Distribution of Petal Lengths',
    edgecolor='white'
)
```

---
## 5. Pie Chart
**What it is:** A circular statistical graphic divided into slices to illustrate numerical proportion.

**When to use:** Use when showing relative proportions or percentages of a total breakdown for a limited number of categories.

```python
species_counts = df['species'].value_with_counts if hasattr(df['species'], 'value_with_counts') else df['species'].value_counts()
species_counts.plot.pie(
    title='Species Proportion in Dataset',
    autopct='%1.1f%%',
    ylabel=''
)
```

---
## 6. Subplot
**What it is:** Arranging multiple distinct plots into grid layout columns and rows within the same master figure.

**When to use:** Use when you want to perform a comparative side-by-side exploration of different variables without blending them onto a single overlapping visual landscape.

```python
df[['sepal_length', 'sepal_width', 'petal_length', 'petal_width']].plot.box(
    subplots=True,
    layout=(2, 2),
    figsize=(10, 8),
    title='Feature Distributions via Subplots'
)
```

---
## 7. Object-Oriented Plotting
**What it is:** Utilizing the underlying target `ax` parameters inside Pandas plotting functions to capture and manipulate specific axes handles.

**When to use:** Use when building production-grade workflows where complex layout compositions, advanced labeling overrides, or multi-layered combinations are mandatory.

```python
# Creating a multi-layered target layout via pandas layout parameters
fig_axes = df['sepal_length'].plot.hist(alpha=0.5, label='Sepal Length', legend=True, title='Multi-layered Target Plot')
df['sepal_width'].plot.hist(alpha=0.5, label='Sepal Width', legend=True, ax=fig_axes)

# Fine-tuning axes handles explicitly directly from pandas-driven configurations
fig_axes.set_xlabel('Centimeters (cm)')
fig_axes.set_ylabel('Observation Frequency')
```
