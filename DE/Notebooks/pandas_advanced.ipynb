```python
ipl_df.columns = ipl_df.columns.str.lower().str.replace('_','')
```

```python
ipl_df['date'] = pd.to_datetime(ipl_df['date'])
```

```python
ipl_df.info()
```

```python
ipl_df.head()
```

```python
ipl_df['season'] = ipl_df['season'].astype('int')
```

```python
ipl_df['tosswinner'].unique()
```

```python
cleanup_map = {
    'Rising Pune Supergiants': 'Rising Pune Supergiant'
}
```

```python
ipl_df['tosswinner'] = ipl_df['tosswinner'].replace(cleanup_map)
```

```python
ipl_df['tosswinner'].unique()
```

```python
ipl_df.head()
```

```python
ipl_df['superover'].unique()
```

```python
bool_map = {
    'N': False,
    'Y': True
}
```

```python
ipl_df['superover'] = ipl_df['superover'].replace(bool_map).fillna(False)
```
