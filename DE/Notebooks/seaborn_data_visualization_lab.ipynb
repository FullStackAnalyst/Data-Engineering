### Initializing the Environment and Dataset
We will load the standard `tips` dataset provided natively by Seaborn to demonstrate these visualizations.

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

# Set standard style
sns.set_theme(style="whitegrid")

# Load dataset
tips = sns.load_dataset("tips")
tips.head()
```

---
## 1. Distplot (Distribution Plot)
**What it is:** A plot that combines a histogram with a kernel density estimate (KDE) to showcase the distribution of a continuous univariate variable.

**When to use:** Use it to check data skewness, kurtosis, and overall underlying data distribution shape. *Note: `sns.distplot` is deprecated in newer versions of Seaborn; `sns.histplot` or `sns.displot` are preferred modern alternatives.*

```python
plt.figure(figsize=(8, 5))
# Using historical distplot with modern fallback warning handled via histplot/kde
sns.histplot(tips['total_bill'], kde=True, color='teal')
plt.title('Distribution of Total Bill')
plt.xlabel('Total Bill ($)')
plt.ylabel('Frequency')
plt.show()
```

---
## 2. Scatterplot
**What it is:** A bivariate plot that displays individual data points as dots across two numeric axes.

**When to use:** Use it to observe relationships, correlations, and clustering trends between two continuous numeric variables, with an option to inject categorical dimensions using colors (`hue`) or markers (`style`).

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=tips, x='total_bill', y='tip', hue='smoker', style='time', palette='Set2')
plt.title('Total Bill vs. Tip Amount')
plt.xlabel('Total Bill ($)')
plt.ylabel('Tip ($)')
plt.show()
```

---
## 3. Joint Plot
**What it is:** A multi-panel figure showing both the bivariate relationship between two variables and their individual univariate distributions along the margins.

**When to use:** Use it when you need an all-in-one look at how two numeric variables interact together while simultaneously inspecting their individual distributions.

```python
g = sns.jointplot(data=tips, x='total_bill', y='tip', kind='scatter', hue='sex', palette='muted')
g.fig.suptitle('Joint Distribution of Total Bill and Tip', y=1.02)
plt.show()
```

---
## 4. Pair Plot
**What it is:** A matrix of pairwise scatterplots across all numeric columns in a dataset, with histograms/KDEs along the diagonal.

**When to use:** Ideal for exploratory data analysis (EDA) when exploring high-dimensional datasets to rapidly identify correlations, outliers, and visual patterns across columns.

```python
sns.pairplot(tips, hue='time', palette='husl')
plt.show()
```

---
## 5. KDE Plot (Kernel Density Estimate)
**What it is:** A non-parametric method to estimate the probability density function of a continuous variable, creating a smooth curved visualization.

**When to use:** Use it when histograms look too jagged or discrete, and you need a cleaner view of continuous data distributions across different categories.

```python
plt.figure(figsize=(8, 5))
sns.kdeplot(data=tips, x='total_bill', hue='time', fill=True, common_norm=False, alpha=0.5, palette='crest')
plt.title('Smooth Density Estimate of Total Bill by Time of Day')
plt.xlabel('Total Bill ($)')
plt.show()
```

---
## 6. Strip Plot
**What it is:** A scatterplot where one variable is categorical. Points are drawn along a single line per category, often with random jitter applied to minimize overlap.

**When to use:** Use it to see the raw data distributions across small discrete groups or categories.

```python
plt.figure(figsize=(8, 5))
sns.stripplot(data=tips, x='day', y='total_bill', hue='smoker', jitter=0.25, dodge=True, palette='Set1')
plt.title('Total Bill Distribution Across Days (Strip Plot)')
plt.xlabel('Day of the Week')
plt.ylabel('Total Bill ($)')
plt.show()
```

---
## 7. Swarm Plot
**What it is:** Similar to a strip plot, but points are non-overlapping and adjusted automatically along the categorical axis to show the exact density shape of the distribution.

**When to use:** Excellent for relatively small datasets where you want to view every single observation clearly without masking distribution density structure.

```python
plt.figure(figsize=(8, 5))
sns.swarmplot(data=tips, x='day', y='total_bill', hue='sex', palette='dark', size=5)
plt.title('Non-overlapping Point Distribution Across Days')
plt.xlabel('Day of the Week')
plt.ylabel('Total Bill ($)')
plt.show()
```

---
## 8. Box Plot
**What it is:** A standardized display of data based on a five-number summary (minimum, first quartile (Q1), median, third quartile (Q3), and maximum), explicitly highlighting outliers.

**When to use:** Ideal for identifying data range, spread, symmetry, and statistical outliers across multiple categorical groups.

```python
plt.figure(figsize=(8, 5))
sns.boxplot(data=tips, x='day', y='total_bill', hue='smoker', palette='vlag')
plt.title('Statistical Summary of Total Bill by Day and Smoking Status')
plt.xlabel('Day of the Week')
plt.ylabel('Total Bill ($)')
plt.show()
```

---
## 9. Violin Plot
**What it is:** A hybrid visualization combining a box plot and a kernel density plot, displaying distribution symmetry and multimodal peaks.

**When to use:** Use it when you want to look at both structural summary statistics (like median) and the underlying data density shapes simultaneously.

```python
plt.figure(figsize=(8, 5))
sns.violinplot(data=tips, x='day', y='total_bill', hue='sex', split=True, inner='quart', palette='pastel')
plt.title('Violin Plot: Total Bill Distribution Shape by Day and Gender')
plt.xlabel('Day of the Week')
plt.ylabel('Total Bill ($)')
plt.show()
```

---
## 10. Bar Plot
**What it is:** A plot that represents estimate central tendencies (usually the mean) using rectangular bars, complete with error bars showing statistical uncertainty.

**When to use:** Use it to compare aggregated metrics (like average value) across different distinct categorical groups.

```python
plt.figure(figsize=(8, 5))
sns.barplot(data=tips, x='day', y='total_bill', hue='sex', errorbar='ci', palette='deep')
plt.title('Average Total Bill Across Days with Confidence Interval Indicators')
plt.xlabel('Day of the Week')
plt.ylabel('Mean Total Bill ($)')
plt.show()
```

---
## 11. Count Plot
**What it is:** A special bar plot variant specifically tailored for counting the frequency of occurrences within categorical groups.

**When to use:** Use it when you want to see the basic sample size distribution or balance of instances across your discrete categorical features.

```python
plt.figure(figsize=(8, 5))
sns.countplot(data=tips, x='day', hue='time', palette='rocket')
plt.title('Sample Volume Count per Day split by Time Slot')
plt.xlabel('Day of the Week')
plt.ylabel('Number of Transactions')
plt.show()
```

---
## 12. Cat Plot (Categorical Plot)
**What it is:** A unified figure-level interface for drawing categorical plots onto a multi-subplot grid mapping rows or columns.

**When to use:** Use it when you want to easily switch between plot types (like box, violin, bar, or swarm) or when you need to break plots out into subplots over columns or rows.

```python
g = sns.catplot(data=tips, x='day', y='total_bill', hue='smoker', col='time', kind='box', palette='YlGnBu', height=4, aspect=1.2)
g.set_axis_labels('Day', 'Total Bill ($)')
plt.show()
```

---
## 13. Heat Map
**What it is:** A 2D visualization representing data magnitudes encoded through color variations over a data matrix.

**When to use:** Commonly used to visualize multi-variable correlation matrices or confusion matrices to evaluate feature interactions rapidly.

```python
plt.figure(figsize=(8, 6))
# Compute numeric correlation
numeric_tips = tips.select_dtypes(include=[np.number])
corr_matrix = numeric_tips.corr()

sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt='.2f', linewidths=.5)
plt.title('Correlation Matrix Heatmap of Numeric Variables')
plt.show()
```

---
## 14. Regression Plot (Regplot)
**What it is:** Fits and plots a linear regression trend line along with a translucent shaded confidence interval band over scatter data points.

**When to use:** Use it when exploring continuous numeric trends to quickly identify if a linear directional dependency exists between two metrics.

```python
plt.figure(figsize=(8, 5))
sns.regplot(data=tips, x='total_bill', y='tip', scatter_kws={'color': 'darkblue', 'alpha':0.6}, line_kws={'color': 'red', 'lw': 2})
plt.title('Linear Regression Trend: Total Bill vs. Tip')
plt.xlabel('Total Bill ($)')
plt.ylabel('Tip ($)')
plt.show()
```
