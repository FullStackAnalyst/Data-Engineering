```python
# Import the visualization and data analysis libraries
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# Load the built-in 'tips' dataset from Seaborn
tips = sns.load_dataset("tips")

# Display the first 5 rows to understand our data structure
tips.head()
```

---
## 1. Scatterplot

### What it is & When to use:
A scatterplot displays individual data points using Cartesian coordinates $(x, y)$.
* **When to use:** Use it to explore the **relationship, correlation, or distribution pattern** between two continuous numeric variables.

In this example, we will see if a higher restaurant bill correlates with a higher tip amount.

```python
# Set the figure size
plt.figure(figsize=(8, 5))

# Plot the data (alpha adds transparency to see overlapping points)
plt.scatter(data=tips, x='total_bill', y='tip', alpha=0.7, color='teal', edgecolor='black')

# Labeling and Formatting
plt.title('Relationship Between Total Bill and Tip Amount', fontsize=14, pad=15)
plt.xlabel('Total Bill ($)', fontsize=12)
plt.ylabel('Tip ($)', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.5)

# Render the plot
plt.show()
```

---
## 2. Line Chart

### What it is & When to use:
A line chart displays information as a series of data points connected by straight line segments.
* **When to use:** Use it to visualize **trends, changes, or trajectories over an ordered sequence** (most commonly time, like days, months, or years).

Because our dataset represents individual transactions rather than a continuous time series, we will group the data by table `size` to observe how the average total bill changes as the party size grows.

```python
# Group data by party size and calculate the average total bill
avg_bill_by_size = tips.groupby('size', observed=False)['total_bill'].mean()

plt.figure(figsize=(8, 5))

# Plot a line chart with markers at each data point
plt.plot(avg_bill_by_size.index, avg_bill_by_size.values, marker='o', color='darkorange', linewidth=2)

# Labeling and Formatting
plt.title('Average Total Bill Trend by Party Size', fontsize=14, pad=15)
plt.xlabel('Party Size (Number of People)', fontsize=12)
plt.ylabel('Average Total Bill ($)', fontsize=12)
plt.grid(True, linestyle=':', alpha=0.6)

plt.show()
```

---
## 3. Bar Chart

### What it is & When to use:
A bar chart represents categorical data with rectangular bars, where the lengths or heights are proportional to the values they represent.
* **When to use:** Use it to **compare distinct groups or discrete categories**.

We will calculate and compare the total revenue (sum of bills) generated across different days of the week.

```python
# Calculate total bill sums grouped by the day of the week
daily_revenue = tips.groupby('day', observed=False)['total_bill'].sum()

plt.figure(figsize=(8, 5))

# Create a bar chart
plt.bar(daily_revenue.index, daily_revenue.values, color='crimson', edgecolor='black', width=0.6)

# Labeling and Formatting
plt.title('Total Revenue Generated by Day of the Week', fontsize=14, pad=15)
plt.xlabel('Day of the Week', fontsize=12)
plt.ylabel('Total Revenue ($)', fontsize=12)

plt.show()
```

---
## 4. Histogram

### What it is & When to use:
A histogram groups continuous numerical data into ranges (called "bins") and counts how many data points fall into each bin.
* **When to use:** Use it to evaluate the **distribution, spread, and frequency** of a single numerical variable (e.g., to look for skewness or outliers).

Let's look at how the tip amounts are distributed to find the most common tipping range.

```python
plt.figure(figsize=(8, 5))

# Plot histogram (bins define the intervals; edgecolor separates bars visually)
plt.hist(tips['tip'], bins=12, color='slateblue', edgecolor='white')

# Labeling and Formatting
plt.title('Distribution of Tip Amounts', fontsize=14, pad=15)
plt.xlabel('Tip Amount ($)', fontsize=12)
plt.ylabel('Frequency (Count)', fontsize=12)

plt.show()
```

---
## 5. Pie Chart

### What it is & When to use:
A pie chart is a circular statistical graphic divided into slices to illustrate numerical proportion.
* **When to use:** Use it exclusively to show **parts-of-a-whole relationships** when dealing with a small number of static categories (ideally fewer than 5).

We will visualize the proportion of dining parties that happened during Lunch vs. Dinner.

```python
# Count the frequency of each meal time
time_counts = tips['time'].value_counts()

plt.figure(figsize=(6, 6))

# Plot pie chart (autopct formats the percentages inside the slices)
plt.pie(time_counts.values, labels=time_counts.index, autopct='%1.1f%%',
        colors=['lightblue', 'gold'], startangle=140, explode=(0, 0.05))

plt.title('Proportion of Dining Transactions: Dinner vs. Lunch', fontsize=14)
plt.show()
```

---
## 6. Object-Oriented Plotting

### What it is & When to use:
Up until now, we used `plt.plot()`, which relies on Matplotlib's global, state-based approach (Pyplot interface). While fast, it becomes messy for complex layouts.

The **Object-Oriented (OO) approach** explicitly creates **Figure** objects (the canvas container) and **Axes** objects (the actual plots/sub-plots).
* **When to use:** This is industry-standard best practice. Use it for **all advanced production code and complex tailoring**, as it gives you explicit control over every graphic element independently.

```python
# Explicitly instantiate the figure and the axis objects
fig, ax = plt.subplots(figsize=(8, 5))

# Use the 'ax' object methods to plot instead of plt
ax.scatter(data=tips, x='total_bill', y='tip', color='forestgreen', alpha=0.6)

# Notice the naming convention change: we use 'set_' for labels and titles in OO style
ax.set_title('Total Bill vs Tip (Object-Oriented Style)', fontsize=14)
ax.set_xlabel('Total Bill ($)', fontsize=12)
ax.set_ylabel('Tip ($)', fontsize=12)
ax.grid(True, alpha=0.3)

plt.show()
```

---
## 7. Subplots

### What it is & When to use:
Subplots allow you to display multiple separate plots arranged in rows and columns within a single Figure canvas.
* **When to use:** Use it when you want to **compare different variables or views of your data side-by-side** without making a single plot overcrowded.

We will use our Object-Oriented framework to generate a 1-row, 2-column subplot panel comparing total bills and tips side-by-side.

```python
# Create a figure matrix grid with 1 row and 2 columns
fig, axes = plt.subplots(nrows=1, ncols=2, figsize=(14, 5))

# --- Plot 1: Total Bill Distribution (Left Axis: Index 0) ---
axes[0].hist(tips['total_bill'], bins=15, color='purple', edgecolor='white')
axes[0].set_title('Distribution of Total Bills', fontsize=12)
axes[0].set_xlabel('Total Bill ($)')
axes[0].set_ylabel('Frequency')

# --- Plot 2: Tip Distribution (Right Axis: Index 1) ---
axes[1].hist(tips['tip'], bins=15, color='indigo', edgecolor='white')
axes[1].set_title('Distribution of Tips', fontsize=12)
axes[1].set_xlabel('Tip Amount ($)')
axes[1].set_ylabel('Frequency')

# Adjust layout automatically so titles/labels don't overlap between plots
plt.tight_layout()

plt.show()
```
