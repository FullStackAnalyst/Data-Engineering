### Initializing the Environment and Dataset
We will load the native `tips` dataset from Seaborn to leverage its numeric and categorical mix for our interactive plots.

```python
import plotly.express as px
import seaborn as sns
import pandas as pd

# Load data
tips = sns.load_dataset("tips")
tips.head()
```

---
## 1. Histogram
**What it is:** A plot that aggregates continuous data into bins to display the underlying frequency distribution.

**When to use:** Use it to quickly analyze the shape, spread, median value, and potential skewness of a single continuous metric.

```python
fig = px.histogram(tips, x="total_bill", nbins=30, title="Distribution of Total Bills", color_discrete_sequence=['teal'])
fig.show()
```

---
## 2. Interactive Scatter Plot
**What it is:** A bivariate point-cloud visualization mapping two quantitative variables along a spatial Cartesian grid.

**When to use:** Use it to investigate core trends, directional correlations, and outliers between numerical columns, using hover states for detailed point inspection.

```python
fig = px.scatter(tips, x="total_bill", y="tip", color="smoker", symbol="time",
                 title="Total Bill vs. Tip (Categorized by Smoking Status & Time)",
                 labels={"total_bill": "Total Bill ($)", "tip": "Tip ($)"})
fig.show()
```

---
## 3. Box Plot
**What it is:** A geometric representation of statistical five-number summaries (minimum, quartiles, median, and maximum) along with concrete outlier parsing.

**When to use:** Ideal for looking at variance, ranges, and checking descriptive distributions across distinct demographic buckets side-by-side.

```python
fig = px.box(tips, x="day", y="total_bill", color="sex", points="all",
             title="Bill Spread Across Operational Days by Gender")
fig.show()
```

---
## 4. Violin Plot
**What it is:** A visualization asset combining summary indicators of box plots alongside computed smooth kernel density curves.

**When to use:** Use it over box plots when dealing with complex, multimodal data groupings where density shape profiles change across target groups.

```python
fig = px.violin(tips, x="day", y="total_bill", color="smoker", box=True, points="all",
                title="Density Shape and Distribution Spread of Total Bills by Day")
fig.show()
```

---
## 5. Bar Chart (Aggregated/Standard)
**What it is:** A chart representing discrete groups using rectangular blocks whose absolute vertical extensions reflect metric weights.

**When to use:** Use it to cleanly illustrate and compare numerical metrics or averages calculated across multiple distinct categories.

```python
fig = px.bar(tips, x="day", y="total_bill", color="time", barmode="group",
             title="Aggregated Total Bill Comparisons by Weekday and Dining Window")
fig.show()
```

---
## 6. Pie / Donut Chart
**What it is:** A circular statistical graphic divided into functional slices to map proportional numerical weights.

**When to use:** Use it strictly to clarify compositional ratios or market share fractions when there are relatively few total category segments.

```python
fig = px.pie(tips, values="total_bill", names="day", hole=0.4,
             title="Proportional Day-wise Contribution to Total Accumulated Revenue")
fig.show()
```

---
## 7. Scatter Matrix (Pair Plot equivalent)
**What it is:** A multi-axis grid framework showing cross-pairwise interaction scatter grids over all combinations of specified data attributes.

**When to use:** Use it during high-dimensional scoping phases to detect macro trends or latent correlations across multiple columns simultaneously.

```python
fig = px.scatter_matrix(tips, dimensions=["total_bill", "tip", "size"], color="time",
                       title="Pairwise Dimensional Matrices Matrix Grouped by Service Window")
fig.show()
```

---
## 8. Density Heatmap
**What it is:** A bivariate distribution framework mapping data concentrations using structured 2D rectangles mapped to a gradient color key.

**When to use:** Use it when dealing with incredibly high-density data matrices where typical standard individual scatter points would overplot and blend into a solid mass.

```python
fig = px.density_heatmap(tips, x="total_bill", y="tip", marginal_x="rug", marginal_y="histogram",
                         title="2D Frequency Density Allocation Map (Total Bill vs. Tip Amount)")
fig.show()
```

---
## 9. Trendline / Regression Fitting
**What it is:** An extension over scatter charts that matches, projects, and overlays mathematical evaluation vectors over underlying variables.

**When to use:** Perfect for validating or communicating suspected linear progressions or directional dynamics between targeted attributes.

```python
fig = px.scatter(tips, x="total_bill", y="tip", trendline="ols", trendline_color_override="darkorange",
                 title="Linear OLS Regression Vector Projections: Bill Weight vs. Realized Tip Metrics")
fig.show()
```

---
## 10. Faceted Plots (Subplots)
**What it is:** Splitting single structural visualization methods into matrix sub-quadrants isolated across underlying specific categorical metrics.

**When to use:** Use it to compare multi-variable distributions cleanly without overcrowding single figures with messy categorical overlapping labels.

```python
fig = px.scatter(tips, x="total_bill", y="tip", facet_col="day", facet_row="time", color="smoker",
                 title="Faceted Analysis Grids Segmented Across Operational Shifts & Days")
fig.show()
```
