```python
data_with_date_type = pd.read_csv(
    "daily-total-female-births.csv", header=0, parse_dates=[0]
)
```

```python
data_with_date_type["Date"].dtype
```

```python
data_with_date_type.shape
```

```python
data_with_date_type[
    (data_with_date_type["Date"] >= "1959-01-01")
    & (data_with_date_type["Date"] <= "1959-01-21")
]
```

```python
data_with_date_type.describe()
```

```python
feature = data_with_date_type.copy()
feature["Year"] = data_with_date_type["Date"].dt.year
feature["Month"] = data_with_date_type["Date"].dt.month
feature["Day"] = data_with_date_type["Date"].dt.day
feature.head()
```

```python
feature["lag1"] = data_with_date_type["Births"].shift(1)
feature["lag2"] = data_with_date_type["Births"].shift(365)
feature.head()
```

```python
feature["Roll_Mean"] = data_with_date_type["Births"].rolling(window=2).mean()
feature.head()
```

```python
feature["Roll_Max"] = data_with_date_type["Births"].rolling(window=3).max()
feature.head()
```

```python
feature["Expand_Max"] = data_with_date_type["Births"].expanding().max()
feature.head()
```
