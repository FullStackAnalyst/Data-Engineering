```python
# format() - Formats specified values in a string

num = 1234567.89123
integer = 255

# < : Left align
print("1. Left Align      : |{:<20}|".format(num))

# > : Right align
print("2. Right Align     : |{:>20}|".format(num))

# ^ : Center align
print("3. Center Align    : |{:^20}|".format(num))

# = : Place sign at leftmost position
print("4. Sign Left       : {:=10}".format(-123))

# + : Always show sign
print("5. Plus Sign       : {:+}".format(123))

# - : Show sign only for negative values
print("6. Minus Sign Only : {:-}".format(123))

# Space : Add leading space for positive values
print("7. Space for +ve   : {: }".format(123))

# , : Comma thousand separator
print("8. Comma Separator : {:,}".format(num))

# _ : Underscore thousand separator
print("9. Underscore Sep. : {:_}".format(num))

# b : Binary
print("10. Binary         : {:b}".format(integer))

# c : Unicode character
print("11. Unicode Char   : {:c}".format(65))

# d : Decimal
print("12. Decimal        : {:d}".format(integer))

# e : Scientific notation (lowercase)
print("13. Scientific e   : {:e}".format(num))

# E : Scientific notation (uppercase)
print("14. Scientific E   : {:E}".format(num))

# f : Fixed-point format
print("15. Fixed Point f  : {:f}".format(num))

# F : Fixed-point format (uppercase INF/NAN)
print("16. Fixed Point F  : {:F}".format(num))

# g : General format
print("17. General g      : {:g}".format(num))

# G : General format (uppercase scientific notation)
print("18. General G      : {:G}".format(num))

# o : Octal
print("19. Octal          : {:o}".format(integer))

# x : Hexadecimal (lowercase)
print("20. Hex Lower      : {:x}".format(integer))

# X : Hexadecimal (uppercase)
print("21. Hex Upper      : {:X}".format(integer))

# n : Locale-aware number format
print("22. Number Format  : {:n}".format(num))

# % : Percentage
print("23. Percentage     : {:%}".format(0.875))
```
