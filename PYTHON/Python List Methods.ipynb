```python
languages = ["Python", "JavaScript", "Java", "C#", "PHP"]
frameworks = ["Django", "React", "Spring Boot", ".NET", "Laravel"]

print(f"Languages: {languages}")
print(f"Frameworks: {frameworks}")
```

```python
# copy()	Returns a copy of the list
languages.copy()
frameworks.copy()

print(f"Languages: {languages}")
print(f"Frameworks: {frameworks}")
```

```python
# append()	Adds an element at the end of the list
languages.append("Ruby")
frameworks.append("Ruby on Rails")

print(f"Languages: {languages}")
print(f"Frameworks: {frameworks}")
```

```python
# count()	Returns the number of elements with the specified value

print(languages.count("Ruby"))
print(frameworks.count("Ruby on Rails"))
```

```python
# index()	Returns the index of the first element with the specified value
print(languages.index("Python"))
print(languages.index("JavaScript"))
print(languages.index("Java"))
print(languages.index("C#"))
print(languages.index("PHP"))
print(languages.index("Ruby"))
print(frameworks.index("Django"))
print(frameworks.index("React"))
print(frameworks.index("Spring Boot"))
print(frameworks.index(".NET"))
print(frameworks.index("Laravel"))
print(frameworks.index("Ruby on Rails"))
```

```python
# insert()	Adds an element at the specified position
languages.insert(1, "Go")
frameworks.insert(1, "Gin")

print(languages)
print(frameworks)
```

```python
# extend()	Add the elements of a list (or any iterable), to the end of the current list
languages.extend(frameworks)

print(languages)
```

```python
# pop()	Removes the element at the specified position
languages.pop(1)

print(languages)
```

```python
# remove()	Removes the item with the specified value
languages.remove("Gin")

print(languages)
```

```python
# reverse()	Reverses the order of the list
languages.reverse()

print(languages)
```

```python
# sort()	Sorts the list

# 1. Ascending (A-Z)
print("\n1. Ascending (A-Z):")
print(sorted(languages))

# 2. Descending (Z-A)
print("\n2. Descending (Z-A):")
print(sorted(languages, reverse=True))

# 3. Sort by Length (Shortest to Longest)
print("\n3. Sort by Length (Shortest to Longest):")
print(sorted(languages, key=len))

# 4. Sort by Length (Longest to Shortest)
print("\n4. Sort by Length (Longest to Shortest):")
print(sorted(languages, key=len, reverse=True))

# 5. Sort by First Character
print("\n5. Sort by First Character:")
print(sorted(languages, key=lambda x: x[0]))

# 6. Sort by Last Character
print("\n6. Sort by Last Character:")
print(sorted(languages, key=lambda x: x[-1]))

# 7. Case-Insensitive Sort
print("\n7. Case-Insensitive Sort:")
print(sorted(languages, key=str.lower))

# 8. Sort by Length, then Alphabetically
print("\n8. Sort by Length, then Alphabetically:")
print(sorted(languages, key=lambda x: (len(x), x)))

# 9. In-Place Ascending Sort
temp = languages.copy()
temp.sort()
print("\n9. In-Place Ascending Sort:")
print(temp)

# 10. In-Place Descending Sort
temp = languages.copy()
temp.sort(reverse=True)
print("\n10. In-Place Descending Sort:")
print(temp)
```

```python
# clear()	Removes all the elements from the list
languages.clear()

print(languages)
```
