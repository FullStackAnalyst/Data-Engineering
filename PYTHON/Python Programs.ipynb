##### How to Add Two Numbers in Python


```python
number1 = int(input("Enter first number: "))
number2 = int(input("Enter second number: "))
```

```python
# Using + Operator
print(number1 + number2)
```

```python
# Using sum() Function
list_of_numbers = [number1, number2]
print(sum(list_of_numbers))
```

```python
# Using operator.add()
import operator

print(operator.add(number1, number2))
```

```python
# Using math.fsum()
import math

print(math.fsum(list_of_numbers))
```

##### Find the Maximum of two numbers in Python


```python
# Using max()
print(max(number1, number2))
```

```python
# Using ternary operator
print(number1 if number1 > number2 else number2)
```

```python
# Using if-Else statement
if number1 > number2:
    print(number1)
else:
    print(number2)
```

##### Factorial of a Number in Python


```python
number = int(input("Enter a number: "))
```

```python
# Using math.factorial()
import math

print(math.factorial(number))
```

```python
# Using NumPy's np.prod()
import numpy as np

if number > 0:
    print(np.prod(range(1, number + 1)))
else:
    print("Factorial is not defined for negative numbers")
```

```python
# Using For Loop
if number < 0:
    print("Factorial is not defined for negative numbers")
else:
    factorial = 1

    for i in range(1, number + 1):
        factorial *= i

    print(factorial)
```

```python
# Using Recursive Function
def factorial(n):
    if n < 0:
        return "Factorial is not defined for negative numbers"
    return 1 if n <= 1 else n * factorial(n - 1)


print(factorial(number))
```

##### Python Program for Simple Interest


```python
principal = float(input("Enter principal amount: "))
rate = float(input("Enter annual interest rate: "))
time = float(input("Enter time: "))
```

```python
# Using function
def simple_interest(principal, rate, time):
    return (principal * rate * time) / 100


print(simple_interest(principal, rate, time))
```

```python
# Using lambda function
simple_interest = lambda p, r, t: (p * r * t) / 100

print(simple_interest(principal, rate, time))
```

```python
# Using list comprehension
simple_interest = [(principal * rate * time) / 100][0]

print(simple_interest)
```
