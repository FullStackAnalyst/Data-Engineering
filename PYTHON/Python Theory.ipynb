##### Python Variables

###### Variable Basics

- Variables are containers used to store data values.
- Variables are created automatically when a value is first assigned to them; Python does not require explicit declarations.
- Variables are dynamically typed, meaning their data types can change during execution.
- Explicit type conversion can be performed using casting.
- The `type()` function returns the data type of a variable.
- String values can be enclosed in either single (`'`) or double (`"`) quotes.
- Variable names can be short (`x`, `y`) or descriptive (`student_name`, `total_volume`).

###### Variable Naming Rules

- A variable name must start with a letter or an underscore (`_`).
- A variable name cannot start with a digit.
- A variable name can contain letters, digits, and underscores only.
- Variable names cannot be Python keywords.
- Variable names are case-sensitive (`age`, `Age`, and `AGE` are different variables).

###### Variable Naming Conventions

- **Camel Case:** The first word starts with a lowercase letter, and subsequent words start with uppercase letters (`studentName`).
- **Pascal Case:** Every word starts with an uppercase letter (`StudentName`).
- **Snake Case:** Words are separated by underscores (`student_name`).
- Descriptive variable names are preferred for better readability.

###### Variable Assignment

- Python allows assigning values to multiple variables in a single statement.
- The same value can be assigned to multiple variables simultaneously.
- Values from iterable objects (such as lists and tuples) can be assigned to individual variables through unpacking.

###### Outputting Variables

- The `print()` function is commonly used to display variable values.
- Multiple variables can be printed by separating them with commas.
- The `+` operator concatenates strings and performs arithmetic addition with numeric values.
- Combining strings and numbers using the `+` operator results in a `TypeError`.
- Using commas in `print()` is the recommended approach when displaying values of different data types.

###### Global and Local Variables

- Variables defined outside functions are global variables and can be accessed throughout the program.
- Variables defined inside functions are local by default and are accessible only within those functions.
- A local variable can have the same name as a global variable without affecting the global variable.
- The `global` keyword is used to create or modify global variables from within a function.

##### Python Strings

- Strings are sequences of characters used to represent textual data.
- String literals can be enclosed in either single quotes (`'`) or double quotes (`"`).
- `'hello'` and `"hello"` represent the same string value.
- The `print()` function can be used to display string literals and string variables.
- Strings can be assigned to variables using the assignment operator (`=`).
- Quotes can be included inside a string, provided they do not match the surrounding quotation marks.
- Multiline strings can be created using triple double quotes (`"""`) or triple single quotes (`'''`).

###### String Slicing

- String slicing is used to extract a portion of a string.
- Slicing uses the syntax: `string[start:end]`.
- The `start` index is inclusive, while the `end` index is exclusive.
- String indexing starts from `0`.
- Omitting the `start` index begins the slice from the beginning of the string.
- Omitting the `end` index extends the slice to the end of the string.
- Negative indices count backward from the end of the string.

###### String Methods

- Python provides several built-in methods for manipulating strings.
- All string methods return new values and do not modify the original string.
- `upper()` converts a string to uppercase.
- `lower()` converts a string to lowercase.
- `strip()` removes leading and trailing whitespace.
- `replace()` replaces occurrences of a specified substring with another substring.
- `split()` divides a string into a list using a specified separator.

###### String Concatenation

- The `+` operator can be used to combine multiple strings.
- Strings and numbers cannot be concatenated directly using the `+` operator.

###### String Formatting

- F-strings are the preferred way to format strings in Python.
- F-strings were introduced in Python 3.6.
- Prefixing a string with `f` enables placeholder expressions using curly braces `{}`.
- Placeholders can contain variables, expressions, function calls, and formatting modifiers.
- Formatting modifiers follow a colon (`:`) inside placeholders.
- The `format()` method can also be used for string formatting.

###### Escape Characters

- Escape characters allow special characters to be included in strings.
- An escape sequence begins with a backslash (`\`).

| Escape Sequence | Meaning           |
| --------------- | ----------------- |
| `\'`            | Single quote      |
| `\"`            | Double quote      |
| `\\`            | Backslash         |
| `\n`            | New line          |
| `\r`            | Carriage return   |
| `\t`            | Horizontal tab    |
| `\b`            | Backspace         |
| `\f`            | Form feed         |
| `\ooo`          | Octal value       |
| `\xhh`          | Hexadecimal value |

##### Python Booleans

- Booleans represent one of two values:
  - `True`
  - `False`
- Boolean values are commonly used in comparisons and decision-making.

###### Boolean Expressions

- Any expression in Python can be evaluated to produce a Boolean result.
- Comparison operations return either `True` or `False`.
- Conditions used in `if` statements evaluate to Boolean values.

###### The `bool()` Function

- The `bool()` function evaluates a value and returns either `True` or `False`.
- Most values evaluate to `True` if they contain data or have non-zero length.

###### Values That Evaluate to `True`

- Non-empty strings
- Non-zero numbers
- Non-empty lists
- Non-empty tuples
- Non-empty sets
- Non-empty dictionaries

###### Values That Evaluate to `False`

- `False`
- `None`
- Numeric zero values, such as `0` and `0.0`
- Empty strings (`""`)
- Empty lists (`[]`)
- Empty tuples (`()`)
- Empty sets (`set()`)
- Empty dictionaries (`{}`)

###### Custom Objects and Boolean Evaluation

- Objects from classes with a `__len__()` method that returns `0` evaluate to `False`.

###### Boolean Functions

- Functions can return Boolean values.
- Boolean-returning functions can be used to control program flow.

###### Built-in Boolean Functions

- Python provides built-in functions that return Boolean values.
- The `isinstance()` function checks whether an object belongs to a specified data type.

###### Boolean and Numeric Equivalence

- `True` behaves like the integer `1`.
- `False` behaves like the integer `0`.


```python
# Boolean Basics
print(True)
print(False)

is_logged_in = True
is_admin = False
```

```python
# Boolean Expressions
print(10 > 5)  # True
print(10 == 5)  # False
print(10 != 5)  # True

x = 18

if x >= 18:
    print("Eligible to vote.")
```

```python
# The bool() Function
print(bool("Python"))  # True
print(bool(100))  # True
print(bool([]))  # False
```

```python
# Values That Evaluate to True
print(bool("Hello"))  # True
print(bool(10))  # True
print(bool([1, 2, 3]))  # True
print(bool((1, 2)))  # True
print(bool({1, 2}))  # True
print(bool({"a": 1}))  # True
```

```python
# Values That Evaluate to False
print(bool(False))  # False
print(bool(None))  # False
print(bool(0))  # False
print(bool(""))  # False
print(bool([]))  # False
print(bool(()))  # False
print(bool(set()))  # False
print(bool({}))  # False
```

```python
# Custom Objects and Boolean Evaluation
class Box:
    def __len__(self):
        return 0


box = Box()

print(bool(box))  # False
```

```python
# Boolean Functions
def is_even(number):
    return number % 2 == 0


print(is_even(4))  # True
print(is_even(7))  # False
```

```python
# Built-in Boolean Functions
print(isinstance(10, int))  # True
print(isinstance("Hello", str))  # True
print(isinstance(3.14, int))  # False
```

```python
# Boolean and Numeric Equivalence
print(True == 1)  # True
print(False == 0)  # True
print(True + True)  # 2
print(False + 5)  # 5
```
